```python

# ============================================================
# Final Report — Cell 1: Collect P1–P5 Results and Figures
# ============================================================

from pathlib import Path
from zipfile import ZipFile, ZIP_DEFLATED
from collections import Counter

from google.colab import drive, files

drive.mount("/content/drive")

project_root = Path(
    "/content/drive/MyDrive/Colab Notebooks/"
    "ML_ECE_571_Project/Code_base_v2"
)

assert project_root.is_dir(), (
    f"Project directory not found: {project_root}"
)

bundle_path = Path(
    "/content/P1_P5_Report_Results_and_Figures.zip"
)

excluded_folders = {
    "Models",
    "Embeddings",
    "Predictions",
    "Checkpoints",
    "__pycache__"
}

selected = []

for folder_name in ("Results", "Figures"):

    folder = project_root / folder_name

    if not folder.exists():
        print(f"[WARNING] Folder not found: {folder}")
        continue

    for path in folder.rglob("*"):

        if not path.is_file():
            continue

        relative_path = path.relative_to(project_root)

        if any(
            part in excluded_folders
            for part in relative_path.parts
        ):
            continue

        suffix = path.suffix.lower()

        if folder_name == "Results":

            allowed = suffix in {
                ".csv", ".json", ".txt",
                ".md", ".yaml", ".yml"
            }

            # Preserve compact saved split manifests.
            if (
                "Manifests" in relative_path.parts
                and suffix == ".npz"
            ):
                allowed = True

        else:

            allowed = suffix == ".pdf"

            # Include PNG only if no same-named PDF exists.
            if suffix == ".png":
                allowed = not path.with_suffix(".pdf").exists()

        if allowed:
            selected.append((path, relative_path))

assert selected, "No report artifacts were found."

with ZipFile(
    bundle_path,
    "w",
    compression=ZIP_DEFLATED,
    compresslevel=6
) as archive:

    for source_path, relative_path in selected:

        archive.write(
            source_path,
            arcname=str(relative_path)
        )

print("=" * 65)
print("FINAL PROJECT REPORT — ARTIFACT COLLECTION")
print("=" * 65)

print(f"Files collected: {len(selected)}")
print(f"ZIP size: {bundle_path.stat().st_size / 1024**2:.2f} MB")

print("\nFile counts:")
for extension, count in sorted(
    Counter(
        path.suffix.lower()
        for path, _ in selected
    ).items()
):
    print(f"  {extension:8s}: {count}")

print("\nSaved:", bundle_path)

files.download(str(bundle_path))
```
